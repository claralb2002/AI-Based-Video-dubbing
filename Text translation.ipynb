## MarianMT  

```python
# pip install sacremoses
# pip install sentencepiece
from transformers import MarianMTModel, MarianTokenizer

text = "The choice of Danish as the target language provides a unique set of linguistic and cultural challenge"
model_name = 'Helsinki-NLP/opus-mt-en-da'
model = MarianMTModel.from_pretrained(model_name)
tokenizer = MarianTokenizer.from_pretrained(model_name)

translated = tokenizer(text, return_tensors="pt", padding=True)
translated_text = model.generate(**translated)
translated_sentence = tokenizer.decode(translated_text[0], skip_special_tokens=True)

print(f"Translated text: {translated_sentence}")
```
