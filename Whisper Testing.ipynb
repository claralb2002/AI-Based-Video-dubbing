# Whisper Testing

## Transcribing from audio file

```python
import whisper

import warnings
# Suppress specific warning
warnings.filterwarnings("ignore", message="FP16 is not supported on CPU; using FP32 instead")
```

```python
# Data
audio_path = "Test_demo.m4a"  
```

```python
# Load the model
model_base = whisper.load_model("base")  # choose: tiny, base, small, medium, large
model_small = whisper.load_model("small")
model_medium = whisper.load_model("medium")
model_large = whisper.load_model("large")
```

```python
result = model_base.transcribe(audio_path)
print(result["text"])
```

```python
result = model_small.transcribe(audio_path)
print(result["text"])
```

```python
result = model_medium.transcribe(audio_path)
print(result["text"])
```

```python
result = model_large.transcribe(audio_path)
print(result["text"])
```
