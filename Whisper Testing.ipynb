## Trancribing live (in chunks)

```python
# load model: base, small, medium, large
model = whisper.load_model("base.en")


def transcribe_audio_in_chunks(audio_file, chunk_duration_ms=5000, model=model):
    """
    Transcribes an audio file in chunks using OpenAI's Whisper.
    
    Parameters:
        audio_file (str): Path to the audio file (supports MP3, M4A, WAV, etc.).
        chunk_duration_ms (int): Duration of each chunk in milliseconds (default: 5000 ms = 5 seconds).
        model_size (str): Whisper model size to use (e.g., "tiny", "base", "small", "medium", "large").
        language (str): Language code to force transcription language (default: "en").
        
    Returns:
        list: A list containing transcriptions for each chunk.
    """

    # Load and convert the audio to 16kHz mono (Whisper's preferred format)
    audio = AudioSegment.from_file(audio_file)
    audio = audio.set_frame_rate(16000).set_channels(1).set_sample_width(2)
    
    transcriptions = []
    num_chunks = len(audio) // chunk_duration_ms + (1 if len(audio) % chunk_duration_ms != 0 else 0)
    
    print("Starting streaming transcription...\n")
    
    # Process each chunk
    for i in range(0, len(audio), chunk_duration_ms):
        chunk = audio[i:i + chunk_duration_ms]
        # Convert chunk to a normalized numpy array (float32 in range [-1, 1])
        samples = np.array(chunk.get_array_of_samples()).astype(np.float32) / 32768.0
        
        # Transcribe the chunk
        result = model.transcribe(samples)
        transcription = result["text"].strip()
        print(f"Chunk {i // chunk_duration_ms + 1}: {transcription}")
        transcriptions.append(transcription)
        
        # Simulate real-time streaming delay
        time.sleep(chunk_duration_ms / 1000.0)
    
    return transcriptions
```
